
## Generate Sales Summary Report

### Objective
Create a text file (`sales_summary.txt`) containing the key findings from the 2026 retail sales dataset.

### Report Contents
1. **Overall Sales Summary**
   - Total sales revenue
   - Total units sold
   - Total transactions
   - Average transaction revenue
   - Average revenue per unit

2. **Product Insights**
   - Highest revenue product
   - Lowest revenue product
   - Highest units sold product

3. **Category Insights**
   - Highest revenue category
   - Lowest revenue category

4. **Monthly Insights**
   - Highest revenue month
   - Lowest revenue month

### Method
- Use previously calculated NumPy results.
- Format the findings using a Python f-string.
- Write the summary to a text file using `open()` and `write()`.

### Output
`reports/sales_summary.txt`


```python
summary = f"""
NUMPY SALES ANALYTICS - 2026
============================

OVERALL SALES SUMMARY
---------------------
Total Revenue: Rs. {total_sales_revenue:,.2f}
Total Units Sold: {total_units_sold}
Total Transactions: {total_transactions}
Average Transaction Revenue: Rs. {revenue.mean():,.2f}
Average Revenue Per Unit: Rs. {average_revenue_per_unit:,.2f}

PRODUCT INSIGHTS
----------------
Highest Revenue Product: {unique_products[product_revenues.argmax()]}
Lowest Revenue Product: {unique_products[product_revenues.argmin()]}
Highest Units Sold Product: {unique_products[product_units.argmax()]}

CATEGORY INSIGHTS
-----------------
Highest Revenue Category: {unique_categories[category_revenues.argmax()]}
Lowest Revenue Category: {unique_categories[category_revenues.argmin()]}

MONTHLY INSIGHTS
----------------
Highest Revenue Month: {months[monthly_revenues.argmax()]}
Lowest Revenue Month: {months[monthly_revenues.argmin()]}
"""

with open("../reports/sales_summary.txt", "w", encoding="utf-8") as file:
    file.write(summary)

print("Sales summary report generated successfully!")
```
